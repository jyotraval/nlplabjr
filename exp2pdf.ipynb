```python
# ==========================================================
# Experiment 2: Working with Text Files and PDF Files in Python
# ==========================================================
# Demonstrates:
#   1. Writing, Reading, Appending Text Files
#   2. Text Statistics (line, word, character count)
#   3. Reading a PDF
#   4. Merging PDFs
#   5. Splitting a PDF
#
# Required library: pypdf
# Install using: pip install pypdf
# ==========================================================

from pypdf import PdfReader, PdfWriter


# ---------------- TEXT FILE OPERATIONS ----------------

def write_text_file():
    # Open file in write mode ('w') - creates new / overwrites
    with open("sample.txt", "w") as file:
        file.write("Hello, Python!\n")
        file.write("Welcome to file handling.")
    print("Text file written successfully.\n")


def read_text_file():
    # Open file in read mode ('r') and read entire content
    with open("sample.txt", "r") as file:
        content = file.read()
    print("File Content:\n" + content + "\n")


def read_line_by_line():
    # Reading file line by line using a for loop
    with open("sample.txt", "r") as file:
        for line in file:
            print(line.strip())
    print()


def append_text_file():
    # Open file in append mode ('a') - adds text, keeps old content
    with open("sample.txt", "a") as file:
        file.write("\nThis is a new line.")
    print("Text appended successfully.\n")


def text_statistics():
    # Count lines, words, and characters in the file
    data = open("sample.txt").read()
    print("Lines :", len(data.splitlines()))
    print("Words :", len(data.split()))
    print("Characters :", len(data))
    print()

```

```python
# ---------------- PDF OPERATIONS ----------------

def read_pdf(pdf_file):
    # Create a PDF reader object
    reader = PdfReader(pdf_file)

    # Print number of pages
    print("Number of pages:", len(reader.pages))

    # Extract text from the first page
    page = reader.pages[0]
    print("Text of first page:\n", page.extract_text(), "\n")


def PDFmerge(pdfs, output):
    # creating pdf file writer object
    pdfWriter = PdfWriter()

    # appending pdfs one by one
    for pdf in pdfs:
        pdfWriter.append(pdf)

    # writing combined pdf to output pdf file
    with open(output, "wb") as f:
        pdfWriter.write(f)

    print("PDFs merged into", output, "\n")


def PDFsplit(pdf, splits):
    # creating pdf reader object
    reader = PdfReader(pdf)

    # starting index of first slice
    start = 0

    for i in range(len(splits) + 1):
        # creating pdf writer object for (i+1)th split
        writer = PdfWriter()

        # output pdf file name
        outputpdf = pdf.split('.pdf')[0] + str(i) + '.pdf'

        try:
            # setting split end position for next split
            end = splits[i]
        except IndexError:
            # setting split end position for last split
            end = len(reader.pages)

        # adding pages to pdf writer object
        for page in range(start, end):
            writer.add_page(reader.pages[page])

        # writing split pdf pages to pdf file
        with open(outputpdf, "wb") as f:
            writer.write(f)

        # interchanging page split start position for next split
        start = end

    print("PDF split successfully.\n")
```

```python
def main():
    while True:
        print("----- MENU -----")
        print("1. Write Text File")
        print("2. Read Text File")
        print("3. Read Text File Line by Line")
        print("4. Append Text File")
        print("5. Text Statistics")
        print("6. Read PDF")
        print("7. Merge PDFs")
        print("8. Split PDF")
        print("9. Exit")
 
        choice = 100
 
        if choice == "1":
            write_text_file()
        elif choice == "2":
            read_text_file()
        elif choice == "3":
            read_line_by_line()
        elif choice == "4":
            append_text_file()
        elif choice == "5":
            text_statistics()
        elif choice == "6":
            pdf_file = input("Enter PDF file name: ")
            read_pdf(pdf_file)
        elif choice == "7":
            pdfs = input("Enter PDF files to merge (comma separated): ").split(",")
            output = input("Enter output file name: ")
            PDFmerge(pdfs, output)
        elif choice == "8":
            pdf = input("Enter PDF file to split: ")
            splits = list(map(int, input("Enter split positions (comma separated): ").split(",")))
            PDFsplit(pdf, splits)
        elif choice == "9":
            print("Exiting program.")
            break
        else:
            print("Invalid choice. Try again.\n")
            break
 
 
# Entry point of the program
if __name__ == "__main__":
    main()
```
