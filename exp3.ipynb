```python
import re
def paragraph_tokenize(text):
    """
    Splits the input text into paragraphs.
    A paragraph is defined as a block of text separated by
    one or more newlines.
    """
    # Split by two or more newlines (handles blank lines between paragraphs)
    paragraphs = re.split(r'\n\s*\n', text.strip())
    # Remove any leading/trailing spaces from each paragraph
    paragraphs = [p.strip() for p in paragraphs if p.strip()]
    return paragraphs

input_text = """
This is the first paragraph.
It has two lines.

This is the second paragraph.

This is the third one, which
also has two lines.
"""

tokens = paragraph_tokenize(input_text)
print("Paragraphs found:", len(tokens))
for i, para in enumerate(tokens, start=1):
    print(f"Paragraph {i}:", para)
```

```python
# ---- Stemming Example ----
import nltk
from nltk.corpus import stopwords
from nltk.stem import LancasterStemmer, PorterStemmer, SnowballStemmer

# Download resources (only once)
# nltk.download('punkt')
# nltk.download('stopwords')
# nltk.download('punkt_tab')  # Download punkt_tab resource

# Sample text
text = "The striped bats are hanging on their feet for best"

# Tokenize text
words = nltk.word_tokenize(text)

# Initialize stemmers
porter = PorterStemmer()
lancaster = LancasterStemmer()
snowball = SnowballStemmer("english")

print("Original Words:", words)

print("\n--- Porter Stemmer ---")
for w in words:
    print(w, "->", porter.stem(w))

print("\n--- Lancaster Stemmer ---")
for w in words:
    print(w, "->", lancaster.stem(w))

print("\n--- Snowball Stemmer ---")
for w in words:
    print(w, "->", snowball.stem(w))
```

```python
# ---- Lemmatization Example (NLTK) ----
import nltk
from nltk.stem import WordNetLemmatizer

# Download WordNet resources (only once)
# nltk.download('punkt')
# nltk.download('wordnet')
# nltk.download('omw-1.4')
# nltk.download('punkt_tab')

# Sample text
text = "The striped bats are hanging on their feet for best"

# Tokenize text
words = nltk.word_tokenize(text)

# Initialize Lemmatizer
lemmatizer = WordNetLemmatizer()

print("Original Words:", words)

print("\n--- Lemmatization (WordNet) ---")
for w in words:
    print(w, "->", lemmatizer.lemmatize(w))  # Default (noun)

print("\n--- Lemmatization with POS tags ---")
for w in words:
    print(
        f"{w} -> Noun: {lemmatizer.lemmatize(w, pos='n')}, Verb: {lemmatizer.lemmatize(w, pos='v')}"
    )
```
